```python
import pandas as pd
import numpy as np
df = pd.read_csv(r"D:\Downloads\titanic\Titanic-Dataset.csv")
print(df.head(5))
print(df.info())
```

```python
df.describe()
```

```python
print(df.dtypes)
```

```python
import matplotlib.pyplot as plt
```

```python
print("numerical attributes: " + df.select_dtypes(include=['int64', 'float64']).columns)
print("categoral attributes: " + df.select_dtypes(include=['object']).columns)
```

```python
#average fare
average_fare = df.groupby('Pclass')['Fare'].mean()

#line plot
plt.plot(average_fare.index, average_fare.values, marker='o')

#title and labels
plt.title("Avg fare by passenger class")
plt.xlabel("passenger class")
plt.ylabel("average fare")

plt.show()
```

```python
#initialise
plt.scatter(df['Age'], df['Fare'])

#title and labels
plt.title("Relationship Bet Passenger age and fare")
plt.xlabel("Age")
plt.ylabel("Fare")

# Display the plot
plt.show()
```

```python
#count passenger gender
gender_count = df['Sex'].value_counts()

#bar chart
plt.bar(gender_count.index, gender_count.values)

#title and labels
plt.title("Number of Male and Female Passengers")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

#display the plot
plt.show()
```

```python
age = df['Age'].value_counts()
plt.hist(age,bins=20)

plt.title("Histogram")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.show()
```

```python
survival_count = df['Survived'].value_counts()

#pie chart
plt.pie(
    survival_count,
    labels=['did Not Survive', 'survived'],
    autopct='%1.1f%%'
)

#title
plt.title("Prop of Passengers Who Survived")

# display
plt.show()
```

```python
# Histogram 
age = df['Age'].dropna()

plt.hist(age, bins=20)

plt.title("Distribution of Passenger Ages")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()

# Count passenger gender
gender_count = df['Sex'].value_counts()

# Bar chart
plt.bar(gender_count.index, gender_count.values)

plt.title("Number of Male and Female Passengers")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
# Create a figure with three subplots
plt.figure(figsize=(15, 5))

# First subplot
plt.subplot(1, 3, 1)

gender_count = df['Sex'].value_counts()

plt.bar(gender_count.index, gender_count.values)
plt.title("Passengers by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

# Second subplot
plt.subplot(1, 3, 2)

plt.hist(df['Fare'], bins=20)
plt.title("Distribution of Passenger Fares")
plt.xlabel("Fare")
plt.ylabel("Frequency")

# Third subplot
plt.subplot(1, 3, 3)

survival_count = df['Survived'].value_counts()

plt.pie(
    survival_count,
    labels=['Did Not Survive', 'Survived'],
    autopct='%1.1f%%'
)

plt.title("Survival Proportion")

# Adjust layout
plt.tight_layout()

# Display the figure
plt.show()
```
